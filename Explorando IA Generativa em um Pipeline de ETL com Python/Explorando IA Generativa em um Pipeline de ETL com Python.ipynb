```python
#Simulando um API de IA para enriquecimento dos dados 

TEMPLATES = [
    "Investir hoje é garantir tranquilidade financeira amanhã.",
    "Pequenos investimentos regulares constroem grandes resultados.",
    "Seu dinheiro pode trabalhar por você com investimentos certos.",
    "Investir é transformar renda em segurança e oportunidades.",
    "Planejamento financeiro começa com bons investimentos."
]

def generate_ai_news(user):
    nome = user.get("Nome", "")
    msg = random.choice(TEMPLATES)

    if nome:
        msg = f"{nome}, {msg}"

    return msg[:100]

for user in users:
    news = generate_ai_news(user)
    print(news)
    user['news'].append(news)
```

```python
df = pd.DataFrame(users)


df['news'] = df['news'].apply(lambda x: ' | '.join(x))


df.to_csv(
    r'C:\Users\Artur\Santander 2025 - Ciência de Dados com Python\dados_clientes_enriquecidos.csv',
    index=False,
    encoding='utf-8-sig'
)
```
